```python
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path

# Constants
kappa = 0.41

# Import measurement files
files = sorted(Path(".").glob("Z_*.dat"))

z = []
u_mean = []

for file in files:

    # Height from filename, e.g. Z_02.7.dat -> z/h = 0.027
    height = float(file.stem.split("_")[1]) / 100 * 20

    # Read measurements
    data = np.loadtxt(file)

    # Extract velocities
    u = data[:, 1]

    # Store mean velocity and height
    z.append(height)
    u_mean.append(np.mean(u))

z = np.array(z)
u_mean = np.array(u_mean)

# Fit logarithmic velocity profile: u = a*ln(z) + b
a, b = np.polyfit(np.log(z), u_mean, 1)

# Calculate shear velocity and roughness length
u_star = kappa * a
z0 = np.exp(-b / a)

print("Shear velocity:", u_star)
print("Roughness length z0/h:", z0)

# Calculate fitted velocity profile
z_fit = np.linspace(min(z), max(z), 200)
u_fit = a * np.log(z_fit) + b

# Plot
plt.figure(figsize=(7, 6))

plt.scatter(u_mean, z, label="Measurements")
plt.plot(u_fit, z_fit, "r-", label="Logarithmic fit")

plt.xlabel("Mean velocity u/U [-]")
plt.ylabel("Height above bed z/h [-]")
plt.title("Logarithmic velocity profile")
plt.legend()
plt.grid(True)
plt.show()

print(height)
```

```python
rho = 1025

tau_w = u_star**2 * rho


print(f'Shear stress from method 1: {tau_w:.2f} Pa')
```

```python
zlist = []
tau_Re = []
for file in files:
    # Height from filename, e.g. Z_02.7.dat -> z/h = 0.027
    height = float(file.stem.split("_")[1]) / 100 * 20 
    data = np.loadtxt(file)
    u = data[:, 1]               
    w = data[:, 2]

    u_prime = u - np.mean(u)     
    w_prime = w - np.mean(w)
    
    tau = rho * -np.mean(u_prime * w_prime)
    tau_Re.append(tau)
    zlist.append(height)

```

```python
z_arr = np.array(zlist)
tau_arr = np.array(tau_Re)
h = 20.0

# Vrije lineaire fit (zonder onderste punt)
a, b = np.polyfit(z_arr[0:14], tau_arr[0:14], 1)
print(f"Vrije fit: tau_b = {b:.2f} Pa, tau=0 bij z = {-b/a:.1f} m")

# Fit met tau = 0 bij z = h opgelegd
x = 1 - z_arr[0:14] / h
tau_b_fixed = np.sum(tau_arr[0:14] * x) / np.sum(x**2)
print(f"Vaste fit: tau_b = {tau_b_fixed:.2f} Pa")

# Plot
zz = np.linspace(0, h, 100)
plt.scatter(tau_arr[0:14], z_arr[0:14], label="Metingen (in fit)")
plt.scatter(tau_arr[0], z_arr[0], color="red", label="Onderste punt (niet in fit)")
plt.plot(a*zz + b, zz, "r--", label="Vrije fit")
plt.plot(tau_b_fixed*(1 - zz/h), zz, "g-", label="Fit met τ=0 bij z=h")
plt.xlabel("τ [Pa]"); plt.ylabel("z [m]")
plt.legend(); plt.grid(True); plt.show()
```
