```python
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path

# Constants
kappa = 0.41

# Import measurement files
files = sorted(Path(".").glob("Z_*.dat"))

z = []
u_mean = []

for file in files:

    # Height from filename, e.g. Z_02.7.dat -> z/h = 0.027
    height = float(file.stem.split("_")[1]) / 100

    # Read measurements
    data = np.loadtxt(file)

    # Extract velocities
    u = data[:, 1]

    # Store mean velocity and height
    z.append(height)
    u_mean.append(np.mean(u))

z = np.array(z)
u_mean = np.array(u_mean)

# Fit logarithmic velocity profile: u = a*ln(z) + b
a, b = np.polyfit(np.log(z), u_mean, 1)

# Calculate shear velocity and roughness length
u_star = kappa * a
z0 = np.exp(-b / a)

print("Shear velocity:", u_star)
print("Roughness length z0/h:", z0)

# Calculate fitted velocity profile
z_fit = np.linspace(min(z), max(z), 200)
u_fit = a * np.log(z_fit) + b

# Plot
plt.figure(figsize=(7, 6))

plt.scatter(u_mean, z, label="Measurements")
plt.plot(u_fit, z_fit, "r-", label="Logarithmic fit")

plt.xlabel("Mean velocity u/U [-]")
plt.ylabel("Height above bed z/h [-]")
plt.title("Logarithmic velocity profile")
plt.legend()
plt.grid(True)
plt.show()
```
